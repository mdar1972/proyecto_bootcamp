## **_1.Introducción_**
La seguridad vial constituye un componente esencial en la planificación urbana y en la promoción de la movilidad sostenible. La siniestralidad vial no solo representa un grave riesgo para la vida de los ciudadanos, sino que también genera elevados costos sociales y económicos que afectan al bienestar colectivo. En el municipio de Envigado, el crecimiento demográfico y el aumento del parque automotor han intensificado las preocupaciones relacionadas con la seguridad vial.

**_El objetivo de este estudio es realizar un análisis exhaustivo de los datos de siniestralidad vial en Envigado, abarcando el período comprendido entre el 1 de enero de 2018 y el 16 de febrero de 2025_**. Mediante el uso de avanzadas técnicas de análisis de datos, se busca identificar patrones de accidentalidad, factores de riesgo y tendencias específicas que permitan desarrollar estrategias eficaces para la prevención de accidentes y la mejora de la movilidad en el municipio.

### **_1.1 Descripción del Problema_**
Los accidentes de tránsito representan una de las principales causas de muerte y lesiones en Colombia. En Envigado, la siniestralidad vial ha mostrado variaciones a lo largo de los años, influenciada por diversos factores como condiciones climáticas, imprudencia del conductor, tipo de vía y franjas horarias críticas. La presente investigación busca realizar un análisis exhaustivo de los datos de siniestralidad vial en el municipio, comprendidos entre el 1 de enero de 2018 y el 16 de febrero de 2025. Este estudio permitirá identificar patrones, tendencias y factores determinantes en la frecuencia y gravedad de los accidentes, con el fin de proponer estrategias de prevención y mitigación.

**Objetivos del Estudio**

1. Identificar las tendencias y patrones de la siniestralidad vial en Envigado.

1. Determinar los factores críticos en la ocurrencia y gravedad de los accidentes.

```python
# Creamos una columna llamada "Franja_Horaria" que permite identificar fácilmente patrones en función del horario del accidente.
# Esto facilitará el análisis
def asignar_franja_horaria(hora):
    if hora >= pd.to_datetime('00:00:00').time() and hora < pd.to_datetime('06:00:00').time():
        return 'Madrugada'
    elif hora >= pd.to_datetime('06:00:00').time() and hora < pd.to_datetime('12:00:00').time():
        return 'Mañana'
    elif hora >= pd.to_datetime('12:00:00').time() and hora < pd.to_datetime('18:00:00').time():
        return 'Tarde'
    else:
        return 'Noche'

# Aplicamos la función para crear la nueva columna 'Franja_Horaria'
df['Franja_Horaria'] = df['Hora_Accidente'].apply(asignar_franja_horaria)

# Verificamos que la nueva columna se haya generado correctamente
print(df[['Hora_Accidente', 'Franja_Horaria']].head(10))
```

```python
# Reemplazamos texto en algunos campos y llenamos con información campos vacíos en otros

# Campo Clase_Vehiculo
df_limpieza['Clase_Vehiculo'] = df_limpieza['Clase_Vehiculo'].replace('Retroexcavadora Rmna', 'Retroexcavadora') # Cambiamos todos los campos que contengan la palabra "Retroexcavadora Rmna" por "Retroexcavadora"
df_limpieza['Clase_Vehiculo'] = df_limpieza['Clase_Vehiculo'].replace('Retroexcavadora Rnma', 'Retroexcavadora') # Cambiamos todos los campos que contengan la palabra "Retroexcavadora Rnma" por "Retroexcavadora"
df_limpieza['Clase_Vehiculo'] = df_limpieza['Clase_Vehiculo'].replace('N.A.', 'Automovil') # Cambiamos todos los campos que contengan la palabra "N.A." por "Retroexcavadora"

# Campo Tipo_Servicio
df_limpieza['Tipo_Servicio'] = df_limpieza['Tipo_Servicio'].replace('Diplomatico', 'Particular') # Cambiamos todos los campos que contengan la palabra "Diplomatico" por "Particular"
df_limpieza['Tipo_Servicio'] = df_limpieza['Tipo_Servicio'].fillna('Particular') # Rellenamos los campos nulos con la palabra "Particular"

# Campo Genero
df_limpieza['Genero'] = df_limpieza['Genero'].replace('Masculino', 'Hombre') # Cambiamos todos los campos que contengan la palabra "Masculino" por "Hombre"
df_limpieza['Genero'] = df_limpieza['Genero'].replace('Femenino', 'Mujer') # Cambiamos todos los campos que contengan la palabra "Femenino" por "Mujer"

# Campo Consumo_Droga
df_limpieza['Consumo_Droga'] = df_limpieza['Consumo_Droga'].replace(0, 'No') # Cambiamos todos los campos que contengan el cero por "No"
df_limpieza['Consumo_Droga'] = df_limpieza['Consumo_Droga'].replace('0', 'No') # Cambiamos todos los campos que contengan el cero como string por "No"
df_limpieza['Consumo_Droga'] = df_limpieza['Consumo_Droga'].fillna('No') # Rellenamos los campos nulos con la palabra "No"
df_limpieza['Consumo_Droga'] = df_limpieza['Consumo_Droga'].replace('S', 'Si') # Cambiamos todos los campos que contengan la letra "S" por "Si"

# Campo Choque_Contra
df_limpieza['Choque_Contra'] = df_limpieza['Choque_Contra'].fillna('No Reportado') # Rellenamos los campos nulos con la palabra "No Reportado"

# Campo Objeto_Fijo
df_limpieza['Objeto_Fijo'] = df_limpieza['Objeto_Fijo'].fillna('Otro Tipo') # Rellenamos los campos nulos con la palabra "Otro Tipo"

# Campo Sector_Accidente
df_limpieza['Sector_Accidente'] = df_limpieza['Sector_Accidente'].replace('No Reporta', 'No Reportado') # Cambiamos todos los campos que contengan la palabra "No Reporta" por "No Reportado"
df_limpieza['Sector_Accidente'] = df_limpieza['Sector_Accidente'].replace(np.nan, 'Rural') # Cambiamos todos los campos nulos por "Rural"
df_limpieza['Sector_Accidente'] = df_limpieza['Sector_Accidente'].fillna('Rural') # Rellenamos los campos nulos con la palabra "Rural"

# Campo Condicion_Climatica
df_limpieza['Condicion_Climatica'] = df_limpieza['Condicion_Climatica'].replace('No Reporta', 'Normal') # Cambiamos todos los campos que contengan la palabra "No Reporta" por "Normal"
df_limpieza['Condicion_Climatica'] = df_limpieza['Condicion_Climatica'].fillna('Normal') # Rellenamos los campos nulos con la palabra "Normal"

# Campo Estado_Embriaguez
df_limpieza['Estado_Embriaguez'] = df_limpieza['Estado_Embriaguez'].replace(0, 'No') # Cambiamos todos los campos que contengan el 0 por "No"
df_limpieza['Estado_Embriaguez'] = df_limpieza['Estado_Embriaguez'].replace(1, 'Si') # Cambiamos todos los campos que contengan el 1 por "Si"
df_limpieza['Estado_Embriaguez'] = df_limpieza['Estado_Embriaguez'].fillna('No') # Rellenamos los campos nulos con la palabra "No"
df_limpieza['Estado_Embriaguez'] = df_limpieza['Estado_Embriaguez'].astype('category') # Cambiamos el campo de tipo float16 a category

# Campo Consumo_Droga
df_limpieza['Consumo_Droga'] = df_limpieza['Consumo_Droga'].replace(0, 'No') # Cambiamos todos los campos que contengan el 0 por "No"
df_limpieza['Consumo_Droga'] = df_limpieza['Consumo_Droga'].replace('S', 'Si') # Cambiamos todos los campos que contengan la "S" por "Si"
df_limpieza['Consumo_Droga'] = df_limpieza['Consumo_Droga'].fillna('No') # Rellenamos los campos nulos con la palabra "No"

# Campo Grado_Embriaguez
# Primero convertimos a tipo string para poder manipular los valores
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].astype(str)

# Reemplazamos los valores
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace('N', 'Negativo') # Remplazamos 'N' por "Negativo"
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace('0', 'Grado 0') # Remplazamos '0' por "Grado 0"
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace(0, 'Grado 0') # Remplazamos 0 por "Grado 0"
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace('1', 'Grado 1')# Remplazamos '1' por "Grado 1"
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace(1, 'Grado 1')# Remplazamos 1 por "Grado 1"
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace('2', 'Grado 2') # Remplazamos '2' por "Grado 2"
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace(2, 'Grado 2') # Remplazamos 2 por "Grado 2"
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace('3', 'Grado 3') # Remplazamos '3' por "Grado 3"
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].replace(3, 'Grado 3') # Remplazamos 3 por "Grado 3"

# Finalmente convertimos a categoría
df_limpieza['Grado_Embriaguez'] = df_limpieza['Grado_Embriaguez'].astype('category') # Cambiamos columna a tipo category
```

```python
# Creamos una copia segura del dataframe original
df_temp = df_unicos.copy()

# Creamos una columna auxiliar para contar incidentes
df_temp['Incidente'] = 1

# Normalizamos Año y Mes en la copia
df_temp['Año'] = df_temp['Año'].astype(str).str.strip()
df_temp['Mes'] = df_temp['Mes'].astype(str).str.strip()

# Convertimos Mes numérico a nombre del mes
meses_dict = {
    '1': 'Enero', '2': 'Febrero', '3': 'Marzo', '4': 'Abril',
    '5': 'Mayo', '6': 'Junio', '7': 'Julio', '8': 'Agosto',
    '9': 'Septiembre', '10': 'Octubre', '11': 'Noviembre', '12': 'Diciembre'
}
df_temp['Mes_Nombre'] = df_temp['Mes'].map(meses_dict)

# Eliminamos filas que no se pudieron mapear correctamente
df_temp = df_temp[df_temp['Mes_Nombre'].notnull()]

# Agrupamos por Año y Mes_Nombre
conteo_final = df_temp.groupby(['Año', 'Mes_Nombre'])['Incidente'].sum().reset_index(name='Cantidad')

# Establecemos el orden cronológico de los meses
orden_meses = [
    'Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio',
    'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'
]
conteo_final['Mes_Nombre'] = pd.Categorical(conteo_final['Mes_Nombre'], categories=orden_meses, ordered=True)

# Creamos la tabla para el heatmap
heatmap_data = conteo_final.pivot_table(index='Año', columns='Mes_Nombre', values='Cantidad', aggfunc='sum')
heatmap_data = heatmap_data.reindex(columns=orden_meses).fillna(0).astype(int)

# Verificamos si hay datos para graficar
if heatmap_data.empty:
    print("⚠️ No hay datos disponibles para generar el mapa de calor.")
else:
    # Generamos el gráfico
    plt.figure(figsize=(12, 6))
    ax = sns.heatmap(
        heatmap_data,
        annot=True,
        fmt='d',
        cmap='YlOrRd',
        linewidths=0.5,
        linecolor='gray',
        cbar_kws={'label': 'Cantidad de Siniestros'}
    )

    ax.set_title('Siniestros por Año y Mes', fontsize=14)
    ax.set_xlabel('Mes')
    ax.set_ylabel('Año')
    plt.figtext(0.5, -0.05, "*El año 2025 contiene datos solo hasta el 16 de febrero", ha='center', fontsize=10)

    plt.tight_layout()
    plt.show()
```

```python
# Cargamos y preparamos los datos de siniestros por año desde df_unicos
df_anio_filtrado = df_unicos.copy()

# Convertimos y limpiamos la columna 'Año'
df_anio_filtrado['Año'] = df_anio_filtrado['Año'].astype(str).str.strip().astype(int)

# Filtramos solo los datos desde 2018 hasta 2024 (excluimos 2025 por ser incompleto)
df_anio_filtrado = df_anio_filtrado[df_anio_filtrado['Año'] <= 2024]

# Creamos columna auxiliar para contar siniestros por año
df_anio_filtrado['Incidente'] = 1

# Agrupamos por año y contamos siniestros
conteo_filtrado = df_anio_filtrado.groupby('Año')['Incidente'].sum().reset_index()
```

```python
# Importamos las librerias para nuestro modelo de regresión lineal y predicción
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# Definimos X (años) y (número de siniestros)
X = conteo_filtrado['Año'].values.reshape(-1, 1)
y = conteo_filtrado['Incidente'].values.reshape(-1, 1)

# Dividimos los datos en entrenamiento (70%) y prueba (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Creamos el modelo y lo entrenamos
modelo_preciso = LinearRegression()
modelo_preciso.fit(X_train, y_train)

# Predicciones en entrenamiento y prueba
y_train_pred = modelo_preciso.predict(X_train)
y_test_pred = modelo_preciso.predict(X_test)

# Evaluamos el desempeño
mse_train = mean_squared_error(y_train, y_train_pred)
r2_train = r2_score(y_train, y_train_pred)
mse_test = mean_squared_error(y_test, y_test_pred)
r2_test = r2_score(y_test, y_test_pred)

# Mostramos métricas
print(f"MSE Entrenamiento: {mse_train:.2f}")
print(f"R² Entrenamiento: {r2_train:.2f}")
print(f"MSE Prueba: {mse_test:.2f}")
print(f"R² Prueba: {r2_test:.2f}")
```

**Desempeño en Entrenamiento**

- MSE (Error Cuadrático Medio): 2530.67

  Este valor indica que, en promedio, las predicciones del modelo tienen una desviación moderada respecto a los datos reales de entrenamiento.

- R² (Coeficiente de Determinación): 0.96

  Un valor excelente. Significa que el modelo logra explicar el 96% de la variabilidad en los datos de entrenamiento. Esto sugiere un ajuste muy preciso sobre los datos que ya conoce.

**Desempeño en Prueba**

- MSE: 189030.86
  
  Este error es drásticamente mayor al del entrenamiento. Las predicciones del modelo sobre los datos de prueba están muy alejadas de los valores reales.

- R²: -6.29
  
  Un valor negativo de R² indica que el modelo no solo no predice bien, sino que es peor que simplemente asumir el promedio de los valores.
Es un claro indicio de sobreajuste (overfitting).

**Diagnóstico**

El modelo está memorando los datos de entrenamiento pero no generaliza bien a nuevos datos.

Esto suele suceder cuando:

- Hay muy pocos datos disponibles (solo 7 años).
- La división en 70/30 genera conjuntos demasiado pequeños para que el modelo aprenda correctamente.
- La relación año-siniestros no sigue una tendencia lineal constante y el modelo lineal no logra capturar otras variaciones.
